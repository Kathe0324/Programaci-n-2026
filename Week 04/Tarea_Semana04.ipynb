# Problema #1: Calculadora de propina inteligente

## Problema

*Crear una calculadora para establecer el total a pagar con la propina incluida*.

## Entrada

*Solicitamos el monto de factura y el % en numeros de la propina que se desea dar*.

## Proceso

- *Una vez que obtenemos los datos (Factura y Porcentaje de propina), iniciamos la formula que seria el total de la factura por el % que se desea dar de propina entre 100, a ese dato lo denominaremos Propina.*

## Salida esperada

- *Una vez que sabemos el valor de Propina lo vamos a sumar al valor de Factura.*

```python
# Escriba la solución aquí
```

```python
try:
  factura = float(input("Ingrese el monto de su factura: "))
  porcentaje = float(input("Ingrese el % en numeros que desea dar de propina: "))

  propina = factura * (float(porcentaje / 100))
  total_a_pagar = factura + propina

  print(f"Total de factura: {factura}")
  print(f"Propina:{propina}")
  print(f"Total a pagar: {total_a_pagar}")

except ValueError:
  print("Error al ingresar datos")
```

# Problema #2: Detector de número par/impar

## Problema

*Clasificar los numeros en 2 grupos (Par e Impar)*

## Entrada

- *Solicitar el numero*

## Proceso

- *Una vez que obtenemos el numero lo divimos entre 2*.

## Salida esperada

- *Si el resultado de esta division es un numero entero significa que el numero es par sino es impar.*

```python
# Escriba la solución aquí
```

```python
try:
  numero = int(input("Ingresa el numero:"))
  if numero % 2 ==0:
    print("El numero que ingresaste es par")
  else:
    print("El numero que ingresaste es impar")
except ValueError:
  print("Debes ingresar un numero valido")

```

# Problema #3: Clasificador de edad

## Problema

*Clasificar la edad en un grupo determinado (Infante, Adolecente, Adulto Joven, Adulto y Adulto Mayor).*

## Entrada

- *Solicitar la edad.*

## Proceso

- *Si es menor a 12 años va en el Grupo Infante.*
- *Si es mayor a 12 años pero menor que 18 años va en el Grupo Adolecente.*
- *Si es mayor de 18 años pero menor de 30 años va en el Grupo Adulto Joven.*
- *Si es mayor de 30 años pero menor de 65 años va en el Grupo Adulto.*
- *Si es mayor de 65 años va en el Grupo Adulto Mayor.*

## Salida esperada

- *Mostrar la clasificacion al grupo designado.*

```python
# Solicita edad y clasifica como menor, adulto o adulto mayor.
```

```python
try:
  edad = int(input("Ingresa tu edad: "))
  if edad <= 12:
    print("¡Genial!, Eres un infante")
  elif edad <= 18:
    print("¡Fantastico!, Eres un adolescente")
  elif edad <= 30:
    print("¡Maravilloso!, Eres un adulto joven")
  elif edad <= 65:
    print("¡Bien!, Eres un adulto")
  else:
    print("¡Excelente!, Eres un adulto mayor")
except ValueError:
  print("Ingresa un numero valido")
```

# Problema #4: Validador de compra con descuento

## Problema

*Solicitar al cliente, monto de la factura y si es cliente frecuente, si es asi se aplica descuento según reglas.*

## Entrada

- *Solicitar el monto de la factura.*
- *Preguntar si es cliente frecuente.*

## Proceso

- *Si es cliente frecuente y el monto de la factura es mayor a 10.000 tiene un 10% de descuento.*
- *Si es cliente frecuente y el monto de la factura es mayor a 15.000 tiene un 15% de descuento.*
- *Si es cliente frecuente y el monto de la factura es mayor a 20.000 tiene un 20% de descuento.*
- *Si no es cliente frecuente no se le aplica ningun descuento.*

## Salida esperada

- *Mostrar el total de factura, el descuento que se le aplicara y el total a pagar.*

## Pruebas manuales


```python
# Escriba la solución aquí
```

```python
try:
  cliente_frecuente = input("Eres cliente frecuente (si/no): ").strip().lower()
  factura = float(input("Ingresa el monto de tu factura: "))
  descuento_porcentaje = 0.0

  if cliente_frecuente == "si":
    if factura >= 20000:
      descuento_porcentaje = 0.20
    elif factura >= 15000:
      descuento_porcentaje = 0.15
    elif factura >= 10000:
      descuento_porcentaje = 0.10

  descuento = factura * descuento_porcentaje
  total_a_pagar = factura - descuento

  print(f"Total de factura: {factura}")
  print(f"Descuento: {descuento}")
  print(f"Total a pagar: {total_a_pagar}")

except ValueError:
  print("Verifique los datos")
```
